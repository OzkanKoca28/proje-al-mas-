```python
# PROJE 7 : Stok Takip Sistemi



ürünler = {}
son_ürün_no = 0

def menü():
    print("---STOK TAKİP SİSTEMİ---")
    print("1. Ürün Ekle")
    print("2. Ürün Sil")
    print("3. Ürün Ara")
    print("4. Ürün Bilgisi Güncelle")
    print("5. Ürünleri Listele")
    print("6. Stok Girişi")
    print("7. Stok Çıkışı")
    print("8. Toplam Ürün Sayısı")
    print("9. Düşük/Yüksek/Tükenmiş Stokları Göster")
    print("10. Kategorileri Göster")
    print("0. Çıkış")



      

def ürün_ekle():
  global son_ürün_no

  ürün_ismi = str(input(f"Lütfen eklemek istediğiniz ürünün ismini giriniz")).capitalize()

  if ürün_ismi in ürünler:
    print(f"{ürün_ismi} ürünü sistemde kayıtlı.")
    print(f"{ürün_ismi} ürününe yapmak istediğiniz işlemi menüden seçiniz.")

  else:
    ürün_kategorisi = input(f"{ürün_ismi} ürünün kategorisini giriniz.")

    if any(harf.isdigit() for harf in ürün_kategorisi):
      print(f"Lütfen kategori isminde rakam kullanmayınız.")

    else:
      try:
        stok_miktarı = int(input(f"{ürün_ismi} ürününün stok sayısını giriniz."))

        if stok_miktarı < 0 :
           print("Stok miktarı negatif olamaz")
           return

      except ValueError:
        print("Lütfen küsüratlı bir değer girmeyiniz...")
        return

      try:
        ürün_fiyatı = int(input(f"{ürün_ismi} ürünün birim fiyatını TL olarak giriniz."))

        if ürün_fiyatı < 0 :
           print("Ürün fiyatı negatif olamaz")
           return
      
      except ValueError:
        print("Lütfen sadece sayı giriniz.")
        return
      

      print("Eklenecek Ürün Bilgileri")
      print(f"Ürünün İsmi: {ürün_ismi}")
      print(f"Ürün Kategorisi: {ürün_kategorisi}")
      print(f"Ürün Stok Miktarı: {stok_miktarı}")
      print(f"Ürün Birim Fiyatı: {ürün_fiyatı} TL")

      hata_düzelt = input("Ürün bilgilerini düzeltmek istiyorsanız EVET istemiyorsanız HAYIR yazınız").upper()

      if hata_düzelt == "EVET":
        düzeltilecek_hata = str(input("Lütfen ürünün düzeltilecek işlemini İSMİ/KATEGORİ/STOK/FİYAT seçeneklerinden birini yazınız")).upper()

        if düzeltilecek_hata == "İSMİ":
          isim = str(input(f"Lütfen ürünün yeni ismini giriniz")).capitalize()
          ürün_ismi = isim

        elif düzeltilecek_hata == "KATEGORİ":
          yeni_kategori = input(f"{ürün_ismi} ürünün yeni kategorisini giriniz.")

          if any(harf.isdigit() for harf in yeni_kategori):
                print(f"Lütfen kategori isminde rakam kullanmayınız.")
          else:
            ürün_kategorisi = yeni_kategori

        elif düzeltilecek_hata == "STOK":
          try:
              yeni_stok_miktarı = int(input(f"{ürün_ismi} ürününün stok sayısını giriniz."))

              if yeni_stok_miktarı < 0:
                 print("Stok miktarı negatif olamaz")
                 return
              else:
               stok_miktarı = yeni_stok_miktarı
          
          except ValueError:
                print("Lütfen küsüratlı bir değer girmeyiniz...")


        elif düzeltilecek_hata == "FİYAT":
          try:
              yeni_ürün_fiyatı = int(input(f"{ürün_ismi} ürünün birim fiyatını TL olarak giriniz."))

              if yeni_ürün_fiyatı < 0: 
                 print("Ürünün fiyatı negatif olamaz")
                 return
              else:
                ürün_fiyatı = yeni_ürün_fiyatı
                  
          except ValueError:
                    print("Lütfen sadece sayı giriniz.")


        else:
            print("Hatalı işlem seçimi! Lüften işlem olarak İSMİ/KATEGORİ/STOK/FİYAT seçiniz")

        # 🌟 EKLEDİĞİMİZ SATIR: Eğer bilgiler değiştiyse, yeni bilgilere göre kodu tekrar üretiyoruz.
        ürün_numarası = ürün_kategorisi[0].upper() + ürün_ismi[0].upper() + str(son_ürün_no)

      else:
          print("Ürün bilgileri sisteme ekleniyor")

      son_ürün_no += 1
      ürün_numarası = ürün_kategorisi[0] + ürün_ismi[0] + str(son_ürün_no)

      ürünler[ürün_numarası] = {
          "İsim": ürün_ismi,
          "Kategori": ürün_kategorisi,
          "Stok": stok_miktarı,
          "Fiyat/TL": ürün_fiyatı
      }
      print(f"{ürün_ismi} başarıyla eklendi!")





def ürün_sil():
   
   silinecek_ürün = str(input("Lütfen silmek istediğiniz ürünün numarasını giriniz.")).upper()

   if silinecek_ürün in ürünler:
      print("Silinecek ürünün bilgileri")
      print(f"Ürün Numarası: {silinecek_ürün}")
      print(f"Ürün İsmi: {ürünler[silinecek_ürün]['İsim']}")
      print(f"Ürünün Kategorisi: {ürünler[silinecek_ürün]['Kategori']}")
      print(f"Ürünün Stok Miktarı: {ürünler[silinecek_ürün]['Stok']}")
      print(f"Ürünün Fiyat: {ürünler[silinecek_ürün]['Fiyat/TL']}")

      seç = input(f"{silinecek_ürün} ürününü silmek istiyorsanız EVET istemiyorsanız HAYIR yazınız.").upper()

      if seç == "EVET":
        del ürünler[silinecek_ürün]
        print(f"{silinecek_ürün} numaralı ürün başarıyla silindi!")

      else:
         print("Silme işlemi iptal ediliyor...")
         return
            
   
   else:
      print("Girdiğiniz numarada ürün bulunmamaktadır.")





def ürün_ara():

  aranacak_ürün_numarası = input("Lütfen aramak istediğiniz ürünün numarasını yazınız.").upper()

  if aranacak_ürün_numarası in ürünler:
     
     print("--- ÜRÜN BİLGİLERİ ---")
     print(f"Ürün Numarası: {aranacak_ürün_numarası}")
     print(f"Ürün İsmi: {ürünler[aranacak_ürün_numarası]['İsim']}")
     print(f"Kategori: {ürünler[aranacak_ürün_numarası]['Kategori']}")
     print(f"Stok: {ürünler[aranacak_ürün_numarası]['Stok']}")
     print(f"Birim Fiyat: {ürünler[aranacak_ürün_numarası]['Fiyat/TL']}")

  else:
     print(f"{aranacak_ürün_numarası} numaralı ürün sistemde bulunmamaktadır.")





def ürün_bilgisi_güncelle():

   güncellenecek_ürün_numarası = input("Lütfen güncellemek istediğiniz ürünün numarasını giriniz.").upper()

   if güncellenecek_ürün_numarası in ürünler:
     
     print("--- ÜRÜNÜN MEVCUT BİLGİLERİ ---")
     print(f"1. Ürün İsmi: {ürünler[güncellenecek_ürün_numarası]['İsim']}")
     print(f"2. Kategori: {ürünler[güncellenecek_ürün_numarası]['Kategori']}")
     print(f"3. Stok: {ürünler[güncellenecek_ürün_numarası]['Stok']}")
     print(f"4. Birim Fiyat/TL: {ürünler[güncellenecek_ürün_numarası]['Fiyat/TL']}")

     güncelleme_seç = input(f"Lütfen güncellemek istediğiniz kayıtın numarasını yazınız.")

     if güncelleme_seç == "1":
        
        yeni_isim = input("Lütfen yeni ismi yazınız.").capitalize()

        onay_i = input(f"Ürünün yeni ismini {yeni_isim} olarak güncellemek istiyormusunuz. Evet/Hayır yazınız ").upper()

        if onay_i == "EVET":
           ürünler[güncellenecek_ürün_numarası]['İsim'] = yeni_isim
           print(f"{güncellenecek_ürün_numarası} Numaralı ürünün ismi {yeni_isim} olarak başarıyla güncellendi.")

        elif onay_i == "HAYIR":
           print("İşlem iptal ediliyor...")
           return

        else:
           print("Lütfen sadece Evet/Hayır yazınız.")
           


     elif güncelleme_seç == "2":
      
      yeni_kategori = input("Lütfen ürünün yeni kategorisini yazınız.").capitalize()

      if any(harf.isdigit() for harf in yeni_kategori):
         print("Kategori isminde rakam bulunamaz.")
         return

      onay_k = input(f"Ürünün yeni kategorisini {yeni_kategori} olarak güncellemek istiyormusunuz. Evet/Hayır yazınız ").upper()

      if onay_k == "EVET":
         ürünler[güncellenecek_ürün_numarası]['Kategori'] = yeni_kategori
         print(f"{güncellenecek_ürün_numarası} Numaralı ürünün kategorisi {yeni_kategori} olarak başarıyla güncellendi.")

      elif onay_k == "HAYIR":
         print("İşlem iptal ediliyor...")
         return

      else:
         print("Lütfen sadece Evet/Hayır yazınız.")


     elif güncelleme_seç == "3":
        
        try:
          yeni_stok = int(input("Lütfen ürünün yeni stok miktarı yazınız."))

          if yeni_stok < 0 :
             print("Ürünün stok miktarı negatif olamaz.")
             return
             
          onay_s = input(f"Ürünün yeni stok miktarı {yeni_stok} olarak güncellemek istiyormusunuz. Evet/Hayır yazınız ").upper()

          if onay_s == "EVET":
             ürünler[güncellenecek_ürün_numarası]['Stok'] = yeni_stok
             print(f"{güncellenecek_ürün_numarası} Numaralı ürünün stok miktarı {yeni_stok} olarak başarıyla güncellendi.")

          elif onay_s == "HAYIR":
             print("İşlem iptal ediliyor...")
             return

          else:
             print("Lütfen sadece Evet/Hayır yazınız.")

        
        except ValueError:
           print("Lütfen küsüratlı değer ve harf girmeyiniz.")
           return

     elif güncelleme_seç == "4":

        try:
           yeni_fiyat =  int(input("Lütfen ürünün yeni birim fiyatını giriniz."))

           if yeni_fiyat < 0 :
              print("Ürünün fiyatı negatif olamaz.")
              return

           onay_f = input(f"Ürünün yeni fiyatını {yeni_fiyat} olarak güncellemek istiyormusunuz. Evet/Hayır yazınız ").upper()

           if onay_f == "EVET":
              ürünler[güncellenecek_ürün_numarası]['Fiyat/TL'] = yeni_fiyat
              print(f"{güncellenecek_ürün_numarası} Numaralı ürünün fiyatı {yeni_fiyat} olarak başarıyla güncellendi.")

           elif onay_f == "HAYIR":
              print("İşlem iptal ediliyor...")
              return

           else:
              print("Lütfen sadece Evet/Hayır yazınız.")
              
  
        except ValueError:
           print("Lütfen sadece rakam giriniz.")
           return

     else:
         print("Lütfen ürünün mevcut bilgileri menüsünden 1,2,3,4 seçeneklerinden birini seçiniz.")

  
   else:
      print(f"{güncellenecek_ürün_numarası} numaralı ürün sistemde bulunmamaktadır.")
      
   



def ürünleri_listele():

  if len(ürünler) : 
  
    print("---ÜRÜN LİSTESİ---\n")

    for ü_no, bilgiler in ürünler.items():
      print(f"Ürün Numarası: {ü_no}")
      print(f"Ürün İsmi: {bilgiler['İsim']}")
      print(f"Ürün Kategorisi: {bilgiler['Kategori']}")
      print(f"Ürün Stok Miktarı: {bilgiler['Stok']}")
      print(f"Ürün Birim Fiyatı: {bilgiler['Fiyat/TL']}\n")
      print("--------------------------------------------\n")

  else:
     print("Ürün listeniz boş")





def stok_girişi():

  if len(ürünler) :

     stok_ürün_no = input("Lütfen stok girişi yapıcağınız ürünün numarasını giriniz.").upper()

     if stok_ürün_no in ürünler:
        try:
           eklenecek_stok = int(input("Lütfen eklemek istediğiniz stok miktarını giriniz."))

           if eklenecek_stok < 0 :
              print("Stok girişi negatif olamaz")
              return

           son_onay = input(f"{eklenecek_stok} Miktarında stok eklemeyi onaylıyorsanız EVET , onaylamıyorsanız HAYIR yazınız").upper()

           if son_onay == "EVET":

              ürünler[stok_ürün_no]['Stok'] += eklenecek_stok
              print(f"{eklenecek_stok} adet stok başarıyla eklendi.")
              print("---GÜNCEL ÜRÜN BİLGİLERİ---")
              print(f"Ürün Numarası: {stok_ürün_no}")
              print(f"Ürün İsmi: {ürünler[stok_ürün_no]['İsim']}")
              print(f"Ürün Kategorisi: {ürünler[stok_ürün_no]['Kategori']}")
              print(f"Ürün Stok Miktarı: {ürünler[stok_ürün_no]['Stok']}")
              print(f"Ürün Birim Fiyatı: {ürünler[stok_ürün_no]['Fiyat/TL']}")


           elif son_onay == "HAYIR":
              print("Stok girişi işlemi iptal ediliyor...")
              return
      
           else:
              print("Lütfen sadece EVET yada HAYIR yazınız.")
      
        except ValueError:
           print("Lütfen sadece rakam giriniz.")

     else:
        print(f"{stok_ürün_no} Numaralı ürün listenizde bulunmamaktadır.")

  else:
     print("Ürün listeniz boş, ilk olarak ürünü listeye ekleyiniz.")
  




def stok_çıkışı():
  
  if len(ürünler) :

     stok_çıkış_ürün_no = input("Lütfen stok çıkışı yapıcağınız ürünün numarasını giriniz.").upper()

     if stok_çıkış_ürün_no in ürünler:
        try:
           çıkarılacak_stok = int(input("Lütfen çıkarmak istediğiniz stok miktarını giriniz."))

           if çıkarılacak_stok < 0 :
              print("Lütfen negatif miktar yazmayınız.")
              return
         
           if çıkarılacak_stok > ürünler[stok_çıkış_ürün_no]['Stok']:
              print("Çıkarılacak stok, mevcut stoktan fazla olamaz.")
              return

           son_çıkış_onay = input(f"{çıkarılacak_stok} Miktarında stok çıkarmayı onaylıyorsanız EVET , onaylamıyorsanız HAYIR yazınız").upper()

           if son_çıkış_onay == "EVET":

              ürünler[stok_çıkış_ürün_no]['Stok'] -= çıkarılacak_stok
              print(f"{çıkarılacak_stok} adet stok başarıyla çıkartıldı.")
              print("---GÜNCEL ÜRÜN BİLGİLERİ---")
              print(f"Ürün Numarası: {stok_çıkış_ürün_no}")
              print(f"Ürün İsmi: {ürünler[stok_çıkış_ürün_no]['İsim']}")
              print(f"Ürün Kategorisi: {ürünler[stok_çıkış_ürün_no]['Kategori']}")
              print(f"Ürün Stok Miktarı: {ürünler[stok_çıkış_ürün_no]['Stok']}")
              print(f"Ürün Birim Fiyatı: {ürünler[stok_çıkış_ürün_no]['Fiyat/TL']}")


           elif son_onay == "HAYIR":
              print("Stok çıkartma işlemi iptal ediliyor...")
              return
      
           else:
              print("Lütfen sadece EVET yada HAYIR yazınız.")
      
        except ValueError:
           print("Lütfen sadece rakam giriniz.")

     else:
        print(f"{stok_çıkış_ürün_no} Numaralı ürün listenizde bulunmamaktadır.")

  else:
     print("Ürün listeniz boş, ilk olarak ürünü listeye ekleyiniz.")





def toplam_ürün_sayısı():

    if ürünler:

        print(f"Toplam ürün miktarı: {len(ürünler)}")

        kategori_sayilari = {}

        for bilgiler in ürünler.values():

            kategori = bilgiler["Kategori"]

            if kategori in kategori_sayilari:
                kategori_sayilari[kategori] += 1

            else:
                kategori_sayilari[kategori] = 1

        print("\n--- KATEGORİLERE GÖRE ÜRÜN SAYILARI ---")

        for kategori, sayı in kategori_sayilari.items():
            print(f"{kategori}: {sayı}")

    else:
        print("Ürün listeniz boş")






def düşük_yüksek_tükenmiş_stokları_göster():
   
   
   if ürünler:
      try:

         belirle = int(input("Lütfen düşük stok sınırı miktarı belirleyiniz"))
         DÜŞÜK_STOK_SINIRI = belirle

         if DÜŞÜK_STOK_SINIRI >= 0:
     
            for ürün_no , bilgiler in ürünler.items():

               # Stok miktarı bitmiş ürünler
               if bilgiler["Stok"] == 0:
                  print("---STOĞU TÜKENMİŞ ÜRÜN BİLGİLERİ---")
                  print(f"Ürün Numarası: {ürün_no}")
                  print(f"Ürün İsmi: {bilgiler['İsim']}")
                  print(f"Ürün Kategori: {bilgiler['Kategori']}")
                  print(f"Ürün Stok Miktarı: {bilgiler['Stok']}")
                  print("---------------------------------------\n")

               # Stok sınırı miktarından düşük olan ürünler
               elif bilgiler['Stok'] <= DÜŞÜK_STOK_SINIRI:
                  print("---DÜŞÜK STOKLU ÜRÜN BİLGİLERİ---")
                  print(f"Ürün Numarası: {ürün_no}")
                  print(f"Ürün İsmi: {bilgiler['İsim']}")
                  print(f"Ürün Kategori: {bilgiler['Kategori']}")
                  print(f"Ürün Stok Miktarı: {bilgiler['Stok']}")
                  print("---------------------------------------\n")

               # Stok sınırı miktarından büyük olan ürünler
               else:
                  print("---YÜKSEK STOKLU ÜRÜN BİLGİLERİ---")
                  print(f"Ürün Numarası: {ürün_no}")
                  print(f"Ürün İsmi: {bilgiler['İsim']}")
                  print(f"Ürün Kategori: {bilgiler['Kategori']}")
                  print(f"Ürün Stok Miktarı: {bilgiler['Stok']}")
                  print("---------------------------------------\n")

         
         else:
            print("Lütfen negatif miktar girmeyiniz")

      except ValueError:
         print("Lütfen sadece sayı giriniz.")
   
   else:
      print("Ürün listeniz boş")


def kategorileri_göster():

   if ürünler:
      print("---KAYITLI KATEGORİLER---")
      
      # 1. Benzersiz kategorileri toplamak için bir 'set' (küme) oluşturuyoruz
      kategoriler = set()
      for bilgiler in ürünler.values():
         kategoriler.add(bilgiler['Kategori'])
      
      # 2. Enumerate ile 1'den başlayarak sıralı bir şekilde yazdırıyoruz
      # start=1 parametresi sayacın 0 yerine 1'den başlamasını sağlar
      for sıra, kategori in enumerate(kategoriler, start=1):
         print(f"{sıra}. {kategori}")

   else:
      print("Ürün listeniz boş")


  
def çıkış():
  print("Stok takip sisteminden çıkış yapılıyor...")



while True:

  menü()

  seçim = input("Lütfen menüden yapmak istediğiniz işlemin numarasını giriniz.")

  if seçim == "1":
    ürün_ekle()

  elif seçim == "2":
    ürün_sil()

  elif seçim == "3":
    ürün_ara()

  elif seçim == "4":
    ürün_bilgisi_güncelle()

  elif seçim == "5":
    ürünleri_listele()

  elif seçim == "6":
    stok_girişi()

  elif seçim == "7":
    stok_çıkışı()

  elif seçim == "8":
    toplam_ürün_sayısı()

  elif seçim == "9":
    düşük_yüksek_tükenmiş_stokları_göster()

  elif seçim == "10":
     kategorileri_göster() 

  elif seçim == "0":
    çıkış()
    break

  else:
    print("Hatalı numara girişi, Lütfen menüdeki işlem numaralarından yapmak istediğiniz işlemin numarasını giriniz.")


```
