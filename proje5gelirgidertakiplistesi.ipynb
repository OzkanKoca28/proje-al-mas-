```python
# PROJE 5 : Gelir-Gider Takip Listesi

gelirler = {}
giderler = {}

def menü():
    print("--GELİR-GİDER MENÜSÜ--")
    print("1. Gelir Ekle")
    print("2. Gider Ekle")
    print("3. Gelirleri Listele")
    print("4. Giderleri Listele")
    print("5. Gelir Sil")
    print("6. Gider Sil")
    print("7. KAR-ZARAR HESAPLA")
    print("8. Toplam Gelir")
    print("9. Toplam Gider")
    print("0. Çıkış")



def gelir_ekle():
    gelir_ismi = input("Lütfen eklemek istediğiniz gelirin ismini yazınız.").capitalize()

    if gelir_ismi in gelirler:
        print(f"{gelir_ismi} isminde geliriniz bulunmaktadır.")
        print(f"{gelir_ismi}: {gelirler[gelir_ismi]}TL")
        seçim = input(f"Listede olan geliri değiştirmek istiyorsanız 'DEĞİŞTİR' , gelirin üzerine eklemek istiyorsanız 'EKLE' yazınız...").upper()

        if seçim == "DEĞİŞTİR":
            try:
                gelir_mik = float(input("Lütfen eklemek istediğiniz gelirin miktarını TL birimi olarak giriniz."))

                if gelir_mik >= 0:   
                    gelirler[gelir_ismi] = gelir_mik
                    print(f"{gelir_ismi} isimli mevcut geliriniz {gelir_mik} TL olarak değiştirildi.")
                else:
                    print("Hata!, Lütfen negatif miktarda gelir girmeyiniz.")

            except ValueError:
                print("Lütfen sadece sayı giriniz.")

        elif seçim == "EKLE":
            try:
                gelir_mik = float(input(("Lütfen eklemek istediğiniz gelirin miktarını TL birimi olarak giriniz.")))
                
                if gelir_mik >= 0:   
                    gelirler[gelir_ismi] += gelir_mik
                    print(f"{gelir_ismi} isimli gelirinizin üzerine {gelir_mik} TL miktarında gelir başarıyla eklenmiştir.")
                    print(f"{gelir_ismi} isimli güncel geliriniz: {gelirler[gelir_ismi]}TL")

                else:
                    print("Hata!, Lütfen negatif miktarda gelir girmeyiniz.")

            except ValueError:
                print("Lütfen sadece sayı giriniz.")

        else:
            print("HATALI SEÇİM!, Lütfen 'DEĞİŞTİR' veya 'EKLE' işleminden birini seçiniz...")

    else: 
         try:
            gelir_mik = float(input("Lütfen eklemek istediğiniz gelirin miktarını TL birimi olarak giriniz."))

            if gelir_mik >= 0:
                gelirler[gelir_ismi] = gelir_mik
                print(f"{gelir_ismi} isimli {gelir_mik} TL miktarında ki geliriniz başarıyla eklenmiştir. ")

            else:
                print("Hata!, Lütfen negatif miktarda gelir girmeyiniz.")
                
         except ValueError:
            print("Lütfen sadece sayı giriniz.")



def gider_ekle():
    gider_ismi = input("Lütfen eklemek istediğiniz giderin ismini yazınız.").capitalize()

    if gider_ismi in giderler:
        print(f"{gider_ismi} isminde gideriniz bulunmaktadır.")
        print(f"{gider_ismi}: {giderler[gider_ismi]}TL")
        seçim = input(f"Listede olan giderinizi değiştirmek istiyorsanız 'DEĞİŞTİR' , giderin üzerine eklemek istiyorsanız 'EKLE' yazınız...").upper()

        if seçim == "DEĞİŞTİR":
            try:
                gider_mik = float(input("Lütfen eklemek istediğiniz giderin miktarını TL birimi olarak giriniz."))

                if gider_mik > 0:   
                    giderler[gider_ismi] = gider_mik
                    print(f"{gider_ismi} isimli mevcut gideriniz {gider_mik} TL olarak değiştirildi.")

                else:
                    print("Hata!, Lütfen negatif miktarda gider girmeyiniz.")

            except ValueError:
                print("Lütfen sadece sayı giriniz.")

        elif seçim == "EKLE":
            try:
                gider_mik = float(input(("Lütfen eklemek istediğiniz giderin miktarını TL birimi olarak giriniz.")))
                
                if gider_mik > 0:   
                    giderler[gider_ismi] += gider_mik
                    print(f"{gider_ismi} isimli giderinizin üzerine {gider_mik} TL miktarında gider başarıyla eklenmiştir.")
                    print(f"{gider_ismi} isimli güncel gideriniz: {giderler[gider_ismi]}TL")

                else:
                    print("Hata!, Lütfen negatif miktarda gider girmeyiniz.")

            except ValueError:
                print("Lütfen sadece sayı giriniz.")

        else:
            print("HATALI SEÇİM!, Lütfen 'DEĞİŞTİR' veya 'EKLE' işleminden birini seçiniz...")

    else: 
        try:
            gider_mik = float(input("Lütfen eklemek istediğiniz giderin miktarını TL birimi olarak giriniz."))

            if gider_mik > 0:
                giderler[gider_ismi] = gider_mik
                print(f"{gider_ismi} isimli {gider_mik} TL miktarında ki gideriniz başarıyla eklenmiştir. ")

            else:
                 print("Hata!, Lütfen negatif miktarda gider girmeyiniz.")


        except ValueError:
            print("Lütfen sadece sayı giriniz.")



def gelirleri_listele():
    if not gelirler :
        print("Listelenecek gelir bulunmamaktadır")

    else:
        print("--- GELİRLER---")

        for sıra, (gelir_ismi, gelir_mik) in enumerate(gelirler.items(), start=1):
        print(f"{sıra}. {gelir_ismi}: {gelir_mik} TL")


        print("------------------")
        print(f"Toplam gelir: {sum(gelirler.values())}")

    

def giderleri_listele():
    if not giderler :
        print("Listelenecek gider bulunmamaktadır")

    else:
        print("--- GİDERLER---")

        for sıra, (gelir_ismi, gelir_mik) in enumerate(gelirler.items(), start=1):
        print(f"{sıra}. {gelir_ismi}: {gelir_mik} TL")

        print("------------------")
        print(f"Toplam gider: {sum(giderler.values())}")



def gelir_sil():
    gelir_ismi = input("Lütfen silmek istediğiniz gelirin ismini giriniz.").capitalize()

    if gelir_ismi not in gelirler :
        print(f"{gelir_ismi} isminde geliriniz bulunmamaktadır.")

    else:
        print(f"{gelir_ismi}: {gelirler[gelir_ismi]}")
        del gelirler[gelir_ismi]
        print(f"{gelir_ismi} ismindeki geliriniz başarıyla silinmiştir.")



def gider_sil():
    gider_ismi = input("Lütfen silmek istediğiniz giderin ismini giriniz.").capitalize()

    if gider_ismi not in giderler :
        print(f"{gider_ismi} isminde gideriniz bulunmamaktadır.")

    else:
        print(f"{gider_ismi}: {giderler[gider_ismi]}")
        del giderler[gider_ismi]
        print(f"{gider_ismi} ismindeki gideriniz başarıyla silinmiştir.")
                


def kâr_zarar_hesapla():
    toplam_gelir = sum(gelirler.values())
    toplam_gider = sum(giderler.values())

    if not gelirler and not giderler :
        print(f"Gelirler ve Giderler listeleriniz boş kar,zarar hesaplanamaz")

    else:
        if toplam_gelir > toplam_gider :
            kâr = toplam_gelir - toplam_gider
            print(f"Toplam gelir: {toplam_gelir}")
            print(f"Toplam gider: {toplam_gider}")
            print(f"Kâr: {kâr}TL")

        elif toplam_gelir == toplam_gider :
            print(f"Toplam gelir: {toplam_gelir}")
            print(f"Toplam gider: {toplam_gider}")
            print(f"Kâr , Zarar : 0 TL")

        else:
            zarar = toplam_gider - toplam_gelir
            print(f"Toplam gelir: {toplam_gelir}")
            print(f"Toplam gider: {toplam_gider}")
            print(f"Zarar: {zarar}TL")

    

def toplam_gelir():
    if not gelirler :
        print("Gelir listenizde hesaplanacak geliriniz bulunmamaktadır.")

    else:
        toplam_gelir = sum(gelirler.values())
        print(f"Toplam gelir: {toplam_gelir}TL")


def toplam_gider():
    if not giderler :
        print("Gider listenizde hesaplanacak gideriniz bulunmamaktadır.")

    else:
        toplam_gider = sum(giderler.values())
        print(f"Toplam gider: {toplam_gider}TL")


def çıkış():
    print("Programdan çıkış yapılıyor...")



while True:
    menü()
    seçim = input("Lütfen menüden yapmak istediğiniz işlemin numarasını giriniz ")

    if seçim == "1":
        gelir_ekle()

    elif seçim == "2":
        gider_ekle()

    elif seçim == "3":
        gelirleri_listele()

    elif seçim == "4":
        giderleri_listele()

    elif seçim == "5":
        gelir_sil()

    elif seçim == "6":
        gider_sil()

    elif seçim == "7":
        kâr_zarar_heapla()

    elif seçim == "8":
        toplam_gelir()

    elif seçim == "9":
        toplam_gider()

    elif seçim == "0":
        çıkış()
        break

    else:
        print("Hatalı numara, Lütfen menüdeki işlem numaralarından giriniz...")
```
