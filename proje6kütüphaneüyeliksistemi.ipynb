```python
# PROJE 6 : Kütüphane Üyelik Sistemi 

üyeler = {}
kitaplar = {
    1: {
        "kitap_adı": "Suç Ve Ceza",
        "yazar": "Dostoyevski",
        "durum": "Rafta"
    },
    2: {
        "kitap_adı": "1984",
        "yazar": "George Orwell",
        "durum": "Rafta"
    },
    3: {
        "kitap_adı": "Sefiller",
        "yazar": "Victor Hugo",
        "durum": "Rafta"
    },
    4: {
        "kitap_adı": "Kürk Mantolu Madonna",
        "yazar": "Sabahattin Ali",
        "durum": "Rafta"
    }
}
son_üye_no = 1

def menü():
    print("---KÜTÜPHANE ÜYELİK SİSTEMİ MENÜSÜ---")
    print("1. Üye Ekle")
    print("2. Üye Sil")
    print("3. Üye Ara")
    print("4. Üye Bilgisi Güncelle")
    print("5. Tüm Üyeleri Listele")
    print("6. Toplam Üye Sayısı")
    print("7. Kitapları Listele")
    print("8. Kitap Teslim Alındı")
    print("9. Kitap Teslim Edildi")
    print("0. Çıkış")


def üye_ekle():
    global son_üye_no
    isim = input("Lütfen eklemek istediğiiniz üyenin ismini giriniz").capitalize()
    
    if any(harf.isdigit() for harf in isim):
        print("İsimde rakam bulunamaz")

    else:
        for üye in üyeler.values():
            if üye["isim"] == isim:
                print(f"{isim} isminde üye zaten kayıtlı.")
                return

        
        telefon_no = input("Eklenecek üyenin telefon numarasını giriniz.")
        telefon_no = telefon_no.replace(" ","")

        if not telefon_no.isdigit():
            print("Telefon numarası sadece rakamlardan oluşmalıdır.")

        
        elif len(telefon_no) != 10:
             print("Telefon numarası 10 haneli olmalıdır.")

        else:
            for üye in üyeler.values():
                if üye["telefon"] == telefon_no:
                    print(f"{telefon_no} telefon numarası başka üyeye ait.")
                    return
                        
            üyeler[son_üye_no]={
                "isim": isim,
                "telefon": telefon_no,
                "emanet": 0,
            }
            print(f"{isim} isimli üye başarıyla eklendi.")
            print(f"Üye numarası: {son_üye_no}")

            son_üye_no += 1





def üye_sil():
    global üyeler

    if not üyeler:
        print("Üye listeniz boş, silinecek üye bulunmamaktadır.")

    else:
        isim = input("Lütfen silmek istediğiniz üyenin ismini giriniz.").capitalize()

        if any(harf.isdigit() for harf in isim):
            print("Lütfen sadece harften oluşan üye ismi giriniz.")

        else:
            for üye_no , üye in üyeler.items():
                if üye["isim"] == isim:
                    print(f"Üye No : {üye_no}")
                    print(f"İsim   : {üye['isim']}")
                    print(f"Telefon: {üye['telefon']}")

                    del üyeler[üye_no]
                     # Üye numaralarını yeniden sırala
                    yeni_üyeler = {}
                    yeni_no = 1

                    for üye in üyeler.values():
                        yeni_üyeler[yeni_no] = üye
                        yeni_no += 1

                    üyeler = yeni_üyeler

                    # Yeni üye numarasını da güncelle
                    global son_üye_no
                    son_üye_no = len(üyeler) + 1

                    print(f"{isim} ismindeki üye başarıyla silinmiştir.")
                    return

            else:
                print(f"{isim} isminde üye bulunmamaktadır.")            






def üye_ara():
    if not üyeler:
        print("Üye listeniz boş, aranacak üye bulunmamaktadır.")

    else:
        isim = input("Lütfen aramak istediğiniz üyenin ismini giriniz.").capitalize()

        if any(harf.isdigit() for harf in isim):
            print("Üye isminde rakam bulunamaz, Lütfen sadece harften oluşan üye ismi giriniz")

        else:
            for üye in üyeler.values():
                if üye["isim"] == isim:
                    print(f"{isim} isimli üyemizin bilgileri aşağıdadır...")
                    print(f"İsim: {üye['isim']}")
                    print(f"Telefon: {üye['telefon']}")
                    print(f"Emanet: {üye['emanet']}")
                    return

            else:
                print(f"{isim} isminde üye bulunmamaktadır.")





def üye_bilgisi_güncelle():
    if not üyeler:
        print("Güncellenecek üye bulunmamaktadır.")

    else:
        isim = input("Güncellemek istediğiniz üyenin ismini giriniz.").capitalize()

        if any(harf.isdigit() for harf in isim):
            print("Üye isminde rakam bulunamaz, Lütfen sadece harften oluşan üye ismi giriniz.")

        else:
            for üye in üyeler.values():
                if üye["isim"] == isim:
                    print("--Mevcut Bilgiler--")
                    print(f"İsim: {üye['isim']}")
                    print(f"Telefon: {üye['telefon']}")
                    print(f"Emanet: {üye['emanet']}")

                    devam_durumu = input(f"Üye bilgisi güncellemeye devam etmek istiyorsanız (EVET), devam etmek istemiyorsanız (HAYIR) yazınız").upper()

                    if devam_durumu == "EVET":

                        print("--Güncelleme İşlemleri--")
                        print("1. İsim Güncelle")
                        print("2. Telefon Bilgisi Güncelle")
                        print("3. Emanet Bilgisi Güncelle")

                        seçim = input("Lütfen güncellemek istediğiniz seçeneğin numarasını giriniz.")

                        if seçim == "1":
                            print(f"Mevcut İsim: {üye['isim']}")
                            yeni_isim = input("Güncellemek istediğiniz yeni ismi yazınız.").capitalize()
                            if any(harf.isdigit() for harf in yeni_isim):
                                print("Girilen isimde rakam bulunmaktadır, Lütfen rakam bulunmayan bir isim giriniz.")
                                return
                            else:
                                üye["isim"] = yeni_isim
                                print(f"İsim {yeni_isim} olarak başarıyla güncellenmiştir.")

                        elif seçim == "2":
                            print(f"Mevcut Telefon: {üye['telefon']}")
                            telefon = input("Lütfen yeni telefon numarasını giriniz.")
                            telefon = telefon.replace(" ","")
                            
                            if not telefon.isdigit():
                                print("Telefon numarası sadece rakamlardan oluşmalıdır.")

                            elif len(telefon) != 10:
                                print("Telefon numarası 10 haneli olmalıdır. Lütfen 10 haneli bir numara giriniz.")

                            else:
                                üye["telefon"] = telefon
                                print(f"Telefon numarası {telefon} numarası olarak başarıyla güncellenmiştir. ")

                        elif seçim == "3":
                            print(f"Mevcut emanet: {üye['emanet']}")
                            emanet = input("Güncellemek istediğiniz yeni emanet numarasını giriniz.")

                            if all(rakam.isdigit() for rakam in emanet):
                                üye["emanet"] = int(emanet)
                                print(f"Emanet sayısı {emanet} olarak başaryıla güncellendi.")

                            else:
                                print("Lütfen sadece rakam giriniz")

                        else:
                            print("Hatalı işlem numarası, Lütfen sadece Güncelleme işlemlerinde bulunan numaralardan birini giriniz...")
                                

                    else:
                        print("Güncelleme işlemi iptal ediliyor")
                        break

            else:
                print(f"{isim} isminde üye bulunmamaktadır.")

        





def tüm_üyeleri_listele():
    if not üyeler: 
        print("Üye listeniz boş, Listelenecek üye bulunmamaktadır.")

    else:
        print("--Üye Listesi--")
        for üye_no , üye in üyeler.items():
            print(f"\n{üye_no}. Üye")
            print(f"İsim: {üye['isim']}")
            print(f"Telefon: {üye['telefon']}")
            print(f"Emanet: {üye['emanet']}")
            print("------------------------------")



def toplam_üye_sayısı():
    if not üyeler:
        print("Üye listeniz boş, Üye sayınız 0(sıfır)")

    else:
        toplam_üye  = len(üyeler.keys())
        print(f"Toplam üye sayısı: {toplam_üye}")



def kitapları_listele():
    print("--Kitaplar--")
    for kitap_no , kitap in kitaplar.items():
        print(f"\n{kitap_no}. Kitap ")
        print(f"Kitap Adı: {kitap['kitap_adı']}")
        print(f"Yazar: {kitap['yazar']}")
        print(f"Durum: {kitap['durum']}")
        print("-------------------------")



def kitap_teslim_alındı():  # Üye kitap ödünç alıyor.
    if not üyeler:
        print("Üye listeniz boş, Kitap teslim alınacak üye bulunmamaktadır.")

    else:
        isim = input("Lütfen eklemek istediğiniz üyenin ismini giriniz.").capitalize()

        if any(harf.isdigit() for harf in isim):
            print("İsimde rakam bulunamaz, Rakam bulunmayan bir isim giriniz.")

        else:
            for üye in üyeler.values():
                if üye["isim"] == isim:
                    kitap_adı_al = input("Lütfen kitap adını giriniz.").title()

                    for kitap in kitaplar.values():
                        if kitap["kitap_adı"] == kitap_adı_al:
                            print(f"--{kitap_adı_al} kitabının durumu--")
                            print(f"Kitap Adı: {kitap['kitap_adı']}")
                            print(f"Yazar: {kitap['yazar']}")
                            print(f"Durum: {kitap['durum']}")

                            if kitap["durum"] == "Rafta":
                                if üye["emanet"] < 3:
                                    seçim = input(f"{kitap_adı_al} ismindeki emanet olarak almak istedğinize eminmisiniz (EVET/HAYIR)").upper()

                                    if seçim == "EVET":
                                        print(f"{kitap_adı_al} kitabı emanet olarak veriliyor...")
                                        üye["emanet"] += 1
                                        kitap["durum"] = "Emanette"
                                        print(f"{kitap_adı_al} kitabı başarıyla teslim edildi.")
                                        print(f"Güncel emanet sayınız: {üye['emanet']}")
                                        return

                                    elif seçim == "HAYIR":
                                        print("Kitap emanet alma işlemi iptal ediliyor...")
                                        return

                                    else:
                                        print(f"Hatalı seçim, lütfen 'EVET/HAYIR' seçeneklerinden birini seçiniz. " )

                                else:
                                    print(f"Kitap emanet sayınız: {üye['emanet']}")
                                    print(f"Kitap emanet alma sınırı 3 olduğundan kitap verilemiyor... ")
                            else:
                                print("Bu kitap şu anda başka bir üyede bulunmaktadır.")
                    else:       
                        print(f"{kitap_adı_al} isminde kitap bulunmamaktadır.")

            else:
                print(f"{isim} isminde üye, listenizde bulunmamaktadır.")
        


def kitap_teslim_edildi(): # Üye ödünç kitabı geri veriyor.
    if not üyeler:
        print("Üye listeniz boş, Kitap teslim almış veya teslim edicek üye bulunmamaktadır. ")

    else:
        isim = input("Lütfen kitap teslim edicek üyenin ismini giriniz.").capitalize()

        if any(harf.isdigit() for harf in isim):
            print("İsimde rakam bulunamaz, Rakam bulunmayan bir isim giriniz.")

        else:

            for üye in üyeler.values():
                if üye["isim"] == isim :
                    kitap_adı_al = input("Lütfen, Teslim edilecek kitabın ismini giriniz.").title()

                    for kitap in kitaplar.values():
                        if kitap["kitap_adı"] == kitap_adı_al:
                            if kitap["durum"] == "Emanette":
                                if üye["emanet"] > 0:

                                    seçim = input("Kitap teslim alınacak (EVET/HAYIR)").upper()

                                    if seçim == "EVET":
                                        print("Kitap teslim alınıyor...")
                                        üye["emanet"] -= 1
                                        kitap["durum"] = "Rafta"
                                        print(f"{kitap_adı_al} kitabını kütüphane başarıyla geri teslim aldı.")
                                        print(f"Güncel emanet sayınız: {üye['emanet']}")
                                        return

                                    elif seçim == "HAYIR":
                                        print("Kitap teslim etme işlemi iptal ediliyor...")
                                        return

                                    else:
                                        print("Hatalı seçim, Lütfen (EVET/HAYIR) seçimlerinden bir tanesini seçiniz.")
                                else:
                                    print("Bu üyenin teslim edeceği emanet kitabı bulunmamaktadır.")
                            else:
                                print(f"Kitap durumu: {kitap['durum']}")
                                print(f"Kitap rafta mevcut, Kitap durumu 'Emanette' olan kitabı geri verebilirsiniz.")
                                return

                    else:
                        print(f"{kitap_adı_al} isminde kitap kaydı bulunmamaktadır.")

            else:
                print(f"{isim} isminde üye, listenizde bulunmamaktadır.")



def çıkış():
    print("Kütüphane Üyelik Sisteminden çıkış yapılıyor...")



while True:
    menü()
    seçim = input("Lütfen menüden yapmak istediğiniz işlemin numarasını giriniz ")

    if seçim == "1":
        üye_ekle()

    elif seçim == "2":
        üye_sil()

    elif seçim == "3":
        üye_ara()

    elif seçim == "4":
        üye_bilgisi_güncelle()

    elif seçim == "5":
        tüm_üyeleri_listele()

    elif seçim == "6":
        toplam_üye_sayısı()

    elif seçim == "7":
        kitapları_listele()

    elif seçim == "8":
        kitap_teslim_alındı()

    elif seçim == "9":
        kitap_teslim_edildi()


    elif seçim == "0":
        çıkış()
        break

    else:
        print("Hatalı numara, Lütfen menüdeki işlem numaralarından giriniz...")    
```
